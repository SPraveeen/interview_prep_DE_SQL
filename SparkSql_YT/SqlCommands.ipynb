# Some major sql commands are
    - DDL -> Data Definition Language
        - Create, drop, truncate, alter, rename
    - DML -> Data manipulation Language
        - Insert, update, merge, delete
    - DCL -> data control Language'
        - Grant, revoke
    - DRL(DQL) -> Data Retrieval (Query) language
        - Select
    - TCL -> Transaction Control Lnaguage
        - Commit, RollBack
