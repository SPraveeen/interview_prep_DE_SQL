### -- /*write a query to find PersonID, Name, number of friends, sum of marks of person who have friends with total score greater than 100. */

```python

with cte1 as (
select 
f.pid as personid,
sum(p.score) as total_score , count(*) as no_of_friends
from person p
inner join friend f 
on p.personid=f.pid
group by personid
having sum(p.score)>100)
select p.PersonId,p.name,no_of_friends,c.total_score as sum_of_marks from cte1 c
inner join person p 
on c.personid=p.personid
```
