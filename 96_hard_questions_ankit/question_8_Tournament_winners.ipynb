-- write a sql query to find the winner in each group

-- the winner in the each group is the player who scored the maximumtotal points within the group. in the case of tie, the lowest palyer_id wins.

```python
create table players
(player_id int,
group_id int)

insert into players values (15,1);
insert into players values (25,1);
insert into players values (30,1);
insert into players values (45,1);
insert into players values (10,2);
insert into players values (35,2);
insert into players values (50,2);
insert into players values (20,3);
insert into players values (40,3);

create table matches
(
match_id int,
first_player int,
second_player int,
first_score int,
second_score int)

insert into matches values (1,15,45,3,0);
insert into matches values (2,30,25,1,2);
insert into matches values (3,30,15,2,0);
insert into matches values (4,40,20,5,2);
insert into matches values (5,35,50,1,1);
```

```python
select * from players;
select * from matches;

with cte1 as (
select d.player,d.score,p.group_id 
from players p inner join (
select first_player as player,first_score as score from matches
union all
select second_player as player,second_score as score from matches) d 
on p.player_id=d.player),cte2 as (select player,group_id,
sum(score)as total_score 
from cte1
group by group_id,player),cte3 as (
select *,dense_rank() over(partition by group_id order by total_score desc)as d_rnk 
from cte2)
select group_id,min(player)as player,total_score from cte3
where d_rnk=1
group by group_id,total_score;
```
