```python
select team_name,COUNT(1) as matches_palyed, sum(win_flag) as no_of_matches_won,count(1)-sum(win_flag) as no_of_losses from (
select team1 as team_name, case when team1=winner then 1 else 0 end as win_flag from table_name
union all 
select team2 as team_name, case when team2=winner then 1 else 0 end as win_flag from table_name) d
group by team_name 
order by no_of_matches_won desc
```
