## python - find duplicate elements in a list

## approach -1

```python
nums=[1,2,3,2,4,5,1,6]
valu=[]
dup_rep=[]

for i in range(len(nums)-1):
    if nums[i] not in valu:
        valu.append(nums[i])
    else:
        dup_rep.append(nums[i])
print(dup_rep)
```

## Approach - 2

```python
d={}

for i in nums:
    if i in d:
        d[i]+=1
    else:
        d[i]=1
print(d)
```

```python
a=[]
for key,value in d.items():
    if value>1:
        a.append(key)
print(a)
```

## find factorial of a number

### with recursive

```python
def find_factorial(n):
    if n==0:
        return 1
    else:
        return n*find_factorial(n-1)
```

```python
print(find_factorial(5))
```

### without recursive

```python
def factorial(n):
    ans=1
    for i in range(1,n+1):
        ans*=i
    return ans
print(factorial(6))
```

## check if a no is prime or not

```python
def perime_no(n):
    if n<=1:
        return False
    for i in range(2,n):
        if n%i==0:
            return False
        return True 
perime_no(12)
```

## second highest distinct salary overall

```python
%sql
select * from (
select emp_id, salary, dense_rank() over(order by salary desc) as salary_rnk from employees) a
where salary_rnk=2
```

## top 3 earners per department with employee_name

```python
%sql
with cte1 as (
select emp_name, dept, dense_rank() over(partition by dept order by salary desc) as salary_rnk from employees)
select * from cte1
where salary_rnk <=3;
```

## flag employees above dept avg salary

```python
%sql
select dept,avg(salary) as avg_salary 
from employees
group by dept
```

```python
%sql
select * from (
select emp_id,emp_name,salary,dept,
avg(salary) over(partition by dept) as avg_salary,
case when salary > avg(salary) over(partition by dept) then true 
else false end as salary_calculator
from employees) a
where salary_calculator=true
```

### JSON String Parsing - Extract values from json

```python
data = [
    (1, '{"product":"Laptop","price":50000,"brand":"Dell"}'),
    (2, '{"product":"Phone","price":30000,"brand":"Samsung"}'),
    (3, '{"product":"Tablet","price":20000,"brand":"Apple"}')
]
df = spark.createDataFrame(data, ["id", "json_str"])
df.display()
```

```python
df.select('id',get_json_object(col('json_str'),'$.product').alias('product'),
               get_json_object(col('json_str'),'$.price').alias('price'),
               get_json_object(col('json_str'),'$.brand').alias('brand')).show()
```
