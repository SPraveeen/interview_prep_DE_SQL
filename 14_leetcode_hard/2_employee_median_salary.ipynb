```python
select company,avg(salary) as avg_salary
from (
select 
*,
row_number() over(partition by company) as rn,
count(1) over(partition by company) as total_cnt 
from employee) d
where rn between total_cnt*1.0/2 and total_cnt*1.0/2 + 1
group by company;
```
